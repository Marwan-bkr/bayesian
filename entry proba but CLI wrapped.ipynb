```python
# -*- coding: utf-8 -*-
import numpy as np
import matplotlib.pyplot as plt
import math
import sys

# --- YOUR ORIGINAL FUNCTIONS (Intact) ---

def probafair(x):
    return 1/6

def probacombifair(x):
    count = 0
    for i in range(1,7):
        for y in range(1,7):
            if i+y == x :
                count = count +1
    return count/(6*6)

def threedoors(choicetype):
    doorsNumber = 3
    proba = 1
    for i in range(doorsNumber-1):
        if choicetype == "stick":
            proba = proba * (1/(doorsNumber))
        elif choicetype == "switch":
            proba = proba * (1/(doorsNumber - i ))
    return proba  # Changed to return for better CLI handling

def annivproblem(npeople):
    day = 365
    diffbirth = 365
    count = 1
    for i in range(npeople - 1):
        day = day - 1
        diffbirth = diffbirth * (day)
    allposs = 365**npeople
    return 1 - diffbirth/allposs

def couponcollect(ntypes):
    return ntypes**ntypes

def myscatter():
    randx = np.linspace(0, 100, 100)
    rand1 = np.random.normal(50, 6, 100)
    rand2 = np.random.normal(56, 2, 100)
    plt.scatter(randx, np.sort(rand1), c="blue", label="Sample 1")
    plt.scatter(randx, np.sort(rand2), c="red", label="Sample 2")
    plt.title("2D Joint Distribution Sample")
    plt.legend()
    plt.show()

    plt.hist(rand2, 30, density=True, color="red", alpha=0.5, label="Var 2")
    plt.hist(rand1, 30, density=True, color="blue", alpha=0.5, label="Var 1")
    plt.title("Marginal Probabilities (Histograms)")
    plt.legend()
    plt.show()

def myPMF():
    randx = np.linspace(0, 10, 100)
    rand1 = np.random.randint(0, 100, 100)
    plt.scatter(randx, rand1, c="blue")
    plt.title("Discrete Distribution Scatter")
    plt.show()

    plt.hist(rand1, 30, density=True, color="blue", alpha=0.5)
    plt.title("PMF (30 Bins)")
    plt.show()

# --- CLI INTERFACE ---

def clear_screen():
    print("\n" * 2)

def main_menu():
    while True:
        clear_screen()
        print("="*40)
        print("   PROBABILITY & STATISTICS TOOLKIT")
        print("="*40)
        print("1. Single Die Roll Probability")
        print("2. Two Dice Sum Probability")
        print("3. Monty Hall Simulator")
        print("4. Birthday Problem Calculator")
        print("5. Coupon Collector Expected Value")
        print("6. Plot 2D Joint Distribution")
        print("7. Plot Discrete PMF")
        print("0. Exit")
        print("-" * 40)

        choice = input("Select an option (0-7): ")

        try:
            if choice == '1':
                num = int(input("Enter die face (1-6): "))
                print(f"\nResult: The probability is {probafair(num):.4f}")

            elif choice == '2':
                target_sum = int(input("Enter the target sum (2-12): "))
                print(f"\nResult: The probability of sum {target_sum} is {probacombifair(target_sum):.4f}")

            elif choice == '3':
                mode = input("Type 'stick' or 'switch': ").lower()
                if mode in ['stick', 'switch']:
                    res = threedoors(mode)
                    print(f"\nResult: Probability of winning by {mode}ing: {res:.4f}")
                else:
                    print("Invalid choice.")

            elif choice == '4':
                people = int(input("Number of people in the room: "))
                res = annivproblem(people)
                print(f"\nResult: Probability of at least two shared birthdays: {res:.4f}")

            elif choice == '5':
                types = int(input("Number of unique coupon types: "))
                print(f"\nResult: {couponcollect(types)}")

            elif choice == '6':
                print("\nGenerating plots... Please close the window to return to menu.")
                myscatter()

            elif choice == '7':
                print("\nGenerating plots... Please close the window to return to menu.")
                myPMF()

            elif choice == '0':
                print("Goodbye!")
                sys.exit()

            else:
                print("Option not recognized.")

        except ValueError:
            print("\nError: Please enter a valid number.")

        input("\nPress Enter to continue...")

if __name__ == "__main__":
    main_menu()
```
