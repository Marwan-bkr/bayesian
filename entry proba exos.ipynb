2. Build a function that models rolling two fair dice and returns the probability that their total equals a chosen sum.

```python
def probacombifair(x):
  count = 0
  for i in range(1,7):
    for y in range(1,7):
      if i+y == x :
        count = count +1
  return count/(6*6)
print(probacombifair(5))
```

3. Simulate the Monty Hall game many times: three doors (one car, two goats), the player picks a door, the host opens another door showing a goat, then the player can keep their door or switch. Compare how often the player wins the car when sticking vs. when switching.

```python
def threedoors(choicetype):
  doorsNumber = 3
  proba = 1
  for i in range(doorsNumber-1):
    if choicetype == "stick":
      proba = proba * (1/(doorsNumber))
    elif choicetype == "switch":
      proba = proba * (1/(doorsNumber - i ))
  print(proba)

threedoors("stick")
```

6. Draw a random sample from a 2D joint distribution, plot the points in a scatter plot, and estimate the marginal probabilities for each variable separately.

```python
import numpy as np
import matplotlib.pyplot as plt

def myscatter():
  randx = np.linspace(0, 100, 100)
  rand1 = np.random.normal(50, 6, 100)
  rand2 = np.random.normal(56, 2, 100)

  plt.scatter(randx, np.sort(rand1), c="blue")
  plt.scatter(randx, np.sort(rand2), c="red")
  plt.show()

  plt.hist(rand2, 30, density=True, color="red", alpha=0.5)
  plt.hist(rand1, 30, density=True, color="blue", alpha=0.5)
  plt.show()

myscatter()
```

7. Build a program that computes and plots the probability mass function (PMF) of a randomly generated discrete distribution.
