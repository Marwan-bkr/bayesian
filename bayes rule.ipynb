# Bayes Rule

```python
import ast  # Safe way to evaluate string inputs as dictionaries

crossproba = {}

def featuresprobainput():
    global crossproba  # Allows us to modify the variable outside the function

    print("Enter the name of feature 1:")
    feature1name = input()
    print("Enter feature 1 categories and weights like {'cat1': 0.6, 'cat2': 0.4}:")
    # We use literal_eval to turn the string input into a real dictionary
    feature1list = ast.literal_eval(input())

    print("Enter the name of feature 2:")
    feature2name = input()
    print("Enter feature 2 categories and weights like {'cat3': 0.3, 'cat4': 0.7}:")
    feature2list = ast.literal_eval(input())

    for cat1 in feature1list:
        for cat2 in feature2list:
            print(f"Enter joint probability of {cat1} and {cat2}:")
            cross_key = f"{cat1}, {cat2}"
            crossproba[cross_key] = input()

def featuresprobaask():
    # Check if the dictionary is not empty
    if crossproba:
        print("\nWhat proba do you want? Enter names of two categories like 'cat1, cat2':")
        askedcategory = input()

        # Using .get() prevents the program from crashing if the user typos
        result = crossproba.get(askedcategory, "Data not found")
        print(f"The probability of {askedcategory} is {result}.")
    else:
        print("Error: No data found. Please run the input function first.")

# Run the functions
featuresprobainput()
featuresprobaask()
```
